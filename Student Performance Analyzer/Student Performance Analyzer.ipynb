### Boolean masking

```python
python_marks = marks[:, 0]
```

```python
print(python_marks > 80)
```

```python
print(marks[python_marks > 80])
```

### Statistics

### Linear Algebra with NumPy

```python
# Vectors
```

```python
import numpy as np
```

```python
a = np.array([2, 4, 6])
b = np.array([1, 3, 5])
print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

```python
np.dot(a, b)
np.linalg.norm(a)
```

```python
# Matrices
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])
```

```python
A + B
```

```python
A - B
```

```python
A * B
```

```python
A @ B
```

```python
# Matrix operations
A.T
np.linalg.det(A)
np.linalg.inv(A)
np.linalg.eig(A)
```
