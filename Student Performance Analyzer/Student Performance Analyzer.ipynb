```python
import numpy as np
```

```python
marks = np.array([
    [85, 78, 92, 88, 76],
    [72, 81, 69, 75, 80],
    [91, 95, 89, 94, 90],
    [60, 65, 58, 70, 62],
    [78, 84, 80, 76, 85]
])
```

```python
print(marks.shape)
```

```python
print(marks.shape[0])
```

```python
print(marks.shape[1])
```

```python
print(marks.sum(axis=1))
```

```python
print(marks.mean(axis=1))
```

```python
print(marks.mean(axis=0))
```

### Indexing and slicing

```python
marks[0]
```

```python
marks[:, 0]
```

```python
marks[:, 2]
```

```python
marks[:3]
```

```python
marks[:, :3]
```

### Boolean masking

```python
python_marks = marks[:, 0]
```

```python
print(python_marks > 80)
```

```python
print(marks[python_marks > 80])
```

### Statistics

```python
np.mean(marks)
```

```python
np.median(marks)
np.std(marks)
```

```python
np.min(marks)
np.max(marks)
```

```python
np.percentile(marks, 25)
np.percentile(marks, 50)
np.percentile(marks, 75)
```

```python
np.mean(marks, axis=0)
np.std(marks, axis=0)
np.min(marks, axis=0)
np.max(marks, axis=0)
```
